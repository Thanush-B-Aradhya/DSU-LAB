```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)

X = np.arange(1, 41)
Y = 2 * X + 1 + np.random.normal(0, 2, 40)


```

```python
data = pd.DataFrame({
    'X': X,
    'Y': Y
})

data

data.to_csv('data.csv', index=False)

print("Dataset created successfully!")
print(data.head())
```
